Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:

    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience


    def describe_employee(self):
        print("Name:", self.name)
        print("Employee Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Year of Experience:", self.year_of_experience)
        print("-----------------------------")

    def update_year_of_experience(self, new_experience):
        self.year_of_experience = new_experience
        print("Experience updated to:", self.year_of_experience)

    # Salary increase karna
    def increase_salary(self, increment):
        self.salary = self.salary + increment
        print("Salary increased to:", self.salary)


employee101 = Employee(
    "Ali",
    101,
    "Karachi",
    100000,
    "Full Stack Developer",
    2022,
    "MERN Stack Diploma",
    3
)

print(employee101.name)
print(employee101.number)
print(employee101.salary)
print(employee101.post)

employee101.describe_employee()

employee101.increase_salary(10000)

employee101.update_year_of_experience(4)


employee102 = Employee("Sara", 102, "Lahore", 90000, "Designer", 2023, "Graphic Design", 2)

employee103 = Employee("Ahmed", 103, "Islamabad", 110000, "Backend Developer", 2021, "BSCS", 4)

employee104 = Employee("Zara", 104, "Karachi", 80000, "Frontend Developer", 2024, "MERN Diploma", 1)

employee105 = Employee("Usman", 105, "Lahore", 95000, "QA Engineer", 2022, "Software Engineering", 3)

employee106 = Employee("Hassan", 106, "Karachi", 130000, "Project Manager", 2020, "MBA", 6)

employee107 = Employee("Ayesha", 107, "Islamabad", 85000, "HR Manager", 2023, "BBA", 2)

employee108 = Employee("Bilal", 108, "Karachi", 105000, "DevOps Engineer", 2021, "Cloud Computing", 4)

employee109 = Employee("Fatima", 109, "Lahore", 75000, "Content Writer", 2024, "English", 1)

employee110 = Employee("Hamza", 110, "Islamabad", 115000, "Mobile Developer", 2022, "Computer Science", 3)

employee101.describe_employee()
employee102.describe_employee()
employee103.describe_employee()
employee104.describe_employee()
employee105.describe_employee()
employee106.describe_employee()
employee107.describe_employee()
employee108.describe_employee()
employee109.describe_employee()
employee110.describe_employee()


employee102.increase_salary(10000)
employee103.update_year_of_experience(5)
employee104.increase_salary(5000)
employee105.update_year_of_experience(4)
employee106.increase_salary(15000)
```
