```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

common = A.intersection(B)

only_a = A.difference(B)

only_b = B.difference(A)

print("Common elements:", common)
print("Only in Set A:", only_a)
print("Only in Set B:", only_b)
```
