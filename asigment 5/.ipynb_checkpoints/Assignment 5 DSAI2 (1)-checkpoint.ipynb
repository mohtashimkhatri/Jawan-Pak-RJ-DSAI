Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

- An inner function calculate_total() → multiplies price * quantity.
- An inner function apply_discount(total) → subtracts discount from total (but never below 0).

The outer function should return the final bill amount.

Expected Output:

- Final Bill = 600 PKR (no discount)
- Final Bill = 500 PKR (with discount)

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for char in text:
        if char in vowels:
            count += 1

    return count


result = count_vowels("Welcome")
print(result)
```

```python
def display_student_info(name, age, grade, school):
    print("Name:", name)
    print("Age:", age)
    print("Grade:", grade)
    print("School:", school)


display_student_info(
    name="John Doe",
    age=15,
    grade="10th",
    school="Greenwood High"
)
```

```python
def display_prices(product, **kwargs):
    print("Product:", product)

    for store, price in kwargs.items():
        print(f"{store}: ${price}")


display_prices(
    "Laptop",
    Amazon=800,
    eBay=750,
    Walmart=780
)
```

```python
def find_max(*args):
    maximum = args[0]

    for num in args:
        if num > maximum:
            maximum = num

    return maximum


result = find_max(10, 25, 5, 80, 40)
print(result)
```

```python
delivery_charges = lambda order_amount: 150 if order_amount < 2000 else 0

print(delivery_charges(1500))
print(delivery_charges(2500))
```

```python
def calculate_rectangle(length, width):

    def calculate_area():
        return length * width

    def calculate_perimeter():
        return 2 * (length + width)

    area = calculate_area()
    perimeter = calculate_perimeter()

    return area, perimeter


area, perimeter = calculate_rectangle(5, 3)

print("Area:", area)
print("Perimeter:", perimeter)
```

```python
minutes_to_seconds = lambda minutes: minutes * 60

result = minutes_to_seconds(5)

print(result)
```

```python
def calculate_hotel_bill(nights):

    def calculate_room_cost():
        return nights * 5000

    def calculate_tax():
        return calculate_room_cost() * 0.10

    room_cost = calculate_room_cost()
    tax = calculate_tax()

    total_bill = room_cost + tax

    return room_cost, tax, total_bill


room_cost, tax, total_bill = calculate_hotel_bill(3)

print("Room Cost:", room_cost)
print("Tax:", tax)
print("Total Bill:", total_bill)
```

```python
def login(username, password):

    def validate_login():
        correct_username = "admin"
        correct_password = "12345"

        if username == correct_username and password == correct_password:
            return "Login Successful"
        else:
            return "Invalid Username or Password"

    return validate_login()


result = login("admin", "12345")

print(result)
```

```python
def shopping_bill(price, quantity, discount=0):

    def calculate_total():
        return price * quantity

    def apply_discount(total):
        return max(0, total - discount)

    total = calculate_total()
    final_bill = apply_discount(total)

    return final_bill


print("Final Bill =", shopping_bill(200, 3), "PKR")

print("Final Bill =", shopping_bill(200, 3, 100), "PKR")
```
