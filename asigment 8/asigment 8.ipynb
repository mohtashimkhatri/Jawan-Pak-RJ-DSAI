```python
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6}
}

cart = {}

while True:
    print("\n1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")

    choice = input("Select an option (1-5): ")

    if choice == "1":
        print("\nAvailable Products:")
        for pid, item in products.items():
            print(f"{pid} - {item['name']} | Rs. {item['price']} | Stock: {item['stock']}")

    elif choice == "2":
        pid = input("Enter Product ID (e.g. P1): ").strip().upper()
        if pid not in products:
            print("Product not found.")
        else:
            qty = int(input("Enter quantity: "))
            if qty <= 0:
                print("Please enter a valid quantity.")
            else:
                current_qty = cart.get(pid, 0)
                if current_qty + qty > products[pid]["stock"]:
                    print(f"Not enough stock. Only {products[pid]['stock']} left.")
                else:
                    cart[pid] = current_qty + qty
                    print(f"{qty} {products[pid]['name']} added to cart.")

    elif choice == "3":
        if not cart:
            print("\nYour cart is empty.")
        else:
            print("\nYour Cart:")
            total = 0
            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                total += subtotal
                print(f"{name} x {qty} = Rs. {subtotal}")
            print(f"Cart Total: Rs. {total}")

    elif choice == "4":
        if not cart:
            print("\nYour cart is empty. Add items before checkout.")
        else:
            total = 0
            print("\nBill Summary:")
            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                total += subtotal
                print(f"{name} x {qty} = Rs. {subtotal}")

            discount = 0
            if total >= 5000:
                discount = total * 0.10

            final_bill = total - discount

            print(f"Subtotal: Rs. {total}")
            if discount > 0:
                print(f"Discount (10%): Rs. {int(discount)}")
            print(f"Final Bill: Rs. {int(final_bill)}")

            for pid, qty in cart.items():
                products[pid]["stock"] -= qty

            cart.clear()
            print("Checkout complete. Thank you for shopping!")

    elif choice == "5":
        print("Exiting program. Goodbye!")
        break

    else:
        print("Invalid option, please try again.")
```
